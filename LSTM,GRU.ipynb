```python
column_to_fill = 'im getting on borderlands and i will murder you all ,'
mode_value = df[column_to_fill].mode()[0]
df[column_to_fill] = df[column_to_fill].fillna(mode_value)
```

```python
df.isnull().sum()
```

# **Data Split**

```python
x=df['im getting on borderlands and i will murder you all ,']
y=df['Positive']
```

```python
print(y.head())
```

```python
print(x.head())
```

# **Label Encoding**

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

```python
print(y)
```

# **Tokenization**

```python
from tensorflow.keras.preprocessing.text import Tokenizer
token=Tokenizer(num_words=5000)
token.fit_on_texts(x)
sq = token.texts_to_sequences(x)
```

# **Padding**

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_pad=pad_sequences(sq,maxlen=20)
```

```python
model = Sequential([

    Embedding(input_dim=5000, output_dim=64),

    LSTM(64),
    Dropout(0.5),

    Dense(4, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.build(input_shape=(None,20))
```

```python
model.summary()
```

```python
print(x_pad.dtype)
print(y.dtype)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x_pad,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model.fit(X_train,y_train,epochs=10,validation_data=(X_test,y_test) )
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print( "Loss :",loss)

print("Accuracy:", accuracy)
```

```python
labels = {
    0: "Irrelevant",
    1: "Negative",
    2: "Neutral",
    3: "Positive"
}
```

```python
text = ["I am  ok"]

seq = token.texts_to_sequences(text)

pad = pad_sequences(seq, maxlen=20)

pred = model.predict(pad)

predicted_class = np.argmax(pred)

print("Predicted Sentiment:",
      labels[predicted_class])
```
